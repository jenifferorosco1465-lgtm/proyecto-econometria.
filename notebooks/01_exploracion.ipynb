```python
import os
import shutil

# Crear la carpeta raw si no existe
os.makedirs("../data/raw", exist_ok=True)

# Archivos que se moverán
archivos = [
    "ECV6R_PERSONAS.SAV",
    "ECV6R_VIVIENDA.SAV",
    "ECV6R_EQUIPAMIENTO.SAV",
    "ECV6R_CREDITOS.SAV",
    "ECV6R_CREDITOS1.SAV",
    "ECV6R_PRESTAMOS.SAV",
    "ECV6R_PRESTAMOS1.SAV",
    "ECV6R_NEGOCIOS.SAV",
    "ECV6R_NEGOCIOS1.SAV",
    "ECV6R_OTROS_INGRESOS.SAV",
    "ECV6R_OTROS_GASTOS.SAV",
    "ECV6R_GASTOS_ALIMENTOS.SAV",
    "ECV6R_GASTOS_ALIMENTOS1.SAV",
    "ECV6R_CAPITAL_SOCIAL.SAV",
    "ECV6R_AGRO1.SAV",
    "ECV6R_AGRO_PARTE_AI.SAV",
    "ECV6R_AGRO_PARTE_AII.SAV",
    "ECV6R_AGRO_PARTE_B.SAV",
    "ECV6R_AGRO_PARTE_C_D.SAV",
    "ECV6R_AGRO_PARTE_E.SAV",
    "ECV6R_AGRO_PARTE_E_IIIA.SAV",
    "ECV6R_AGRO_PARTE_E_IV.SAV",
    "ECV6R_AGRO_PARTE_E_V_F_G.SAV",
    "ecv_fexp_fin.sav",
    "Nota aclaratoria.pdf"
]

for archivo in archivos:
    origen = os.path.join("../data", archivo)
    destino = os.path.join("../data/raw", archivo)
    if os.path.exists(origen):
        shutil.move(origen, destino)
        print(f"Movido: {archivo}")

print("✅ Archivos organizados correctamente.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Seleccionar variables
modelo = personas[
    ["trabaja", "SEXO", "EDAD", "PE47", "AREA_5000", "FEXP"]
].copy()

# Eliminar personas sin información necesaria
modelo = modelo.dropna()

print("Observaciones:", len(modelo))
modelo.head()
```

```python
# Crear variable hombre
modelo["hombre"] = (modelo["SEXO"].astype(str) == "Hombre").astype(int)

# Convertir EDAD a numérica
modelo["EDAD"] = pd.to_numeric(modelo["EDAD"].astype(str), errors="coerce")

# Crear edad al cuadrado
modelo["edad2"] = modelo["EDAD"] ** 2

print(modelo[["trabaja", "hombre", "EDAD", "edad2", "PE47", "AREA_5000"]].head())
```

```python
print("Observaciones:", len(modelo))

print("\nSituación laboral:")
print(modelo["trabaja"].value_counts())

print("\nSexo:")
print(modelo["SEXO"].value_counts())

print("\nÁrea:")
print(modelo["AREA_5000"].value_counts())

print("\nEducación:")
print(modelo["PE47"].value_counts())
```

```python
# Convertir variables categóricas en variables dummy
modelo = pd.get_dummies(
    modelo,
    columns=["PE47", "AREA_5000"],
    drop_first=True,
    dtype=int
)

print("Número de observaciones:", modelo.shape[0])
print("Número de variables:", modelo.shape[1])

print("\nVariables del modelo:")
print(modelo.columns.tolist())
```
